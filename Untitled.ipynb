```python
import cv2
cap = cv2.VideoCapture(1)
print("Press 's' to save image, 'q' to quit.")
while True:
    ret, frame = cap.read()
    if not ret:
        break
    cv2.imshow("Shelf Checkerboard Capture", frame)
    key = cv2.waitKey(1)
    if key == ord('s'):
        cv2.imwrite("shelf_checkerboard.jpg", frame)
        print("Saved as self_checkerboard.jpg")
        break
        if key == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()



```
