```python
import cv2
import numpy as np

CHECKERBOARD = (9, 6)
square_size = 0.025  # 25 mm

camera_matrix = np.load("camera_matrix.npy")
dist_coeffs = np.load("dist_coeffs.npy")

# 3D checkerboard points
objp = np.zeros((CHECKERBOARD[0] * CHECKERBOARD[1], 3), np.float32)
objp[:, :2] = np.mgrid[0:CHECKERBOARD[0], 0:CHECKERBOARD[1]].T.reshape(-1, 2)
objp *= square_size

# Read one image of the checkerboard ON THE SHELF
img = cv2.imread("shelf_checkerboard.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

ret, corners = cv2.findChessboardCorners(gray, CHECKERBOARD, None)

if not ret:
    raise Exception("Checkerboard not found.")

# Solve camera pose
ret, rvec, tvec = cv2.solvePnP(objp, corners, camera_matrix, dist_coeffs)

# Convert to matrices
R, _ = cv2.Rodrigues(rvec)

np.save("R_shelf.npy", R)
np.save("T_shelf.npy", tvec)

print("Extrinsic calibration done.")
print("R =\n", R)
print("T =\n", tvec)
```
