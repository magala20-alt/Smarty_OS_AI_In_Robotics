```python
import requests

PI_SERVER_URL = "http://<PI-IP>:5050/find"   # Example: http://192.168.1.14:5050/find

def send_command_to_pi(command):
    """Send command to Pi and wait for detection results."""

    # Extract only the animal name\
    command = command.lower()

    animals = ["tiger", "lion", "cheetah", "giraffe", "elephant", "zebra"]
    target = None

    for a in animals:
        if a in command:
            target = a
            break

    if not target:
        return {"status": "error", "message": "Unknown animal"}

    try:
        response = requests.post(PI_SERVER_URL, json={"animal": target})
        return response.json()
    except:
        return {"status": "error", "message": "Pi unreachable"}
```
